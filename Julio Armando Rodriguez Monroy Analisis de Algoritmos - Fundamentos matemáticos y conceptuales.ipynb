**RESPUESTA:**

El **Principio de Invarianza** establece que el tiempo de ejecución de dos implementaciones distintas de un mismo algoritmo **no va a diferir más que en una constante multiplicativa**.

- *Define qué es un 'algoritmo', asegurándote de mencionar sus características clave, y describe qué significa el 'Problema de Paro'formulado por Alan Turing.*- *Sintetiza la diferencia entre realizar un 'Análisis a Priori' y una 'Prueba a Posteriori' para medir la eficiencia de un algoritmo.*  
**RESPUESTA:**

Un algoritmo es una secuencia ordenada, lógica y finita de pasos o instrucciones precisas que se siguen para resolver un problema o realizar una tarea. En la vida cotidiana, es muy similar a una receta de cocina: si sigues las instrucciones en el orden correcto paso a paso, obtienes el plato deseado.
Para que un conjunto de instrucciones sea verdaderamente un algoritmo, debe cumplir con las siguientes características clave:
Entrada (Input): Es la información o los datos iniciales que el algoritmo necesita para empezar a trabajar.
Salida (Output): Es el resultado final o la solución que produce el algoritmo una vez que termina su proceso.
No ambiguo (Preciso): Cada instrucción debe ser completamente clara y explícita. No puede haber dudas ni dejar nada a la interpretación; incluso una computadora (que no piensa por sí misma) debe poder seguir los pasos al pie de la letra.
Finito: El algoritmo tiene que terminar en algún momento después de un número limitado de pasos. Un proceso que se queda repitiéndose para siempre en un ciclo infinito no es un algoritmo.
Correcto: El algoritmo debe resolver el problema de manera correcta y entregar la salida adecuada para cualquier dato de entrada válido.
Efectivo (y determinista): Cada paso debe ser lo suficientemente básico como para poder realizarse en un tiempo razonable. Además, si ejecutas el algoritmo varias veces con los mismos datos de entrada, siempre debes obtener exactamente el mismo resultado.
General: Debe servir para resolver cualquier variante del problema general (por ejemplo, ordenar cualquier lista de números) y no solo un caso particular.
2. ¿Qué es el "Problema de Paro" de Alan Turing?
Formulado por el brillante matemático Alan Turing en 1936, el Problema de Paro (o Halting Problem) es uno de los pilares de la informática teórica. El problema plantea la siguiente pregunta:
¿Es posible diseñar un programa de computadora universal que, al analizar cualquier otro programa de computadora y sus datos de entrada, sea capaz de decirnos con absoluta certeza si ese programa eventualmente terminará (parará) o si se quedará corriendo para siempre en un ciclo infinito?.
Turing demostró de forma matemática que este problema es indecidible. Esto significa que es imposible crear un algoritmo universal que resuelva este problema para todos los casos.
Aunque para algunos programas sencillos sí podamos predecir si van a parar o no, Turing probó que siempre existirán programas con lógicas tan complejas o contradictorias que ningún analizador automático podrá descifrar. Tampoco sirve la opción de "simplemente correr el programa y esperar", porque si el programa cae en un ciclo infinito, nos quedaríamos esperando eternamente sin saber nunca si iba a terminar en el siguiente segundo o nunca.
3. Diferencia entre "Análisis a Priori" y "Prueba a Posteriori"
Para medir qué tan eficiente es un algoritmo (es decir, cuánto tiempo tarda o cuánta memoria consume), se utilizan dos enfoques o fases de análisis muy distintas:
Análisis A Priori (Teórico)
¿Cómo se hace?: Se realiza en papel, antes de escribir el código o programar, estudiando la representación lógica o el pseudocódigo del algoritmo. Se calcula matemáticamente una función que define cómo crecerá el tiempo de ejecución a medida que aumente el volumen de los datos de entrada.
Ventaja clave: Es totalmente independiente de la computadora, del lenguaje de programación o de la habilidad del programador. Te permite saber si un algoritmo es óptimo desde la etapa de diseño, ahorrándote el trabajo de programar un método lento o ineficiente.
Prueba A Posteriori (Empírica o Experimental)
¿Cómo se hace?: Se realiza después de programar y ejecutar el algoritmo en una computadora real. Consiste en realizar pruebas de laboratorio ejecutando el programa con datos reales de muestra y midiendo con un reloj interno (o estadísticas del sistema) el tiempo exacto en milisegundos y los recursos de memoria consumidos.
Limitación clave: Los resultados dependen por completo de factores externos al algoritmo, como la potencia del hardware, el sistema operativo, la calidad del compilador y la habilidad de quien programó el código. Además, solo mide el comportamiento para un conjunto limitado de pruebas y no garantiza cómo responderá el programa con otros datos.
Sintetizando la diferencia: El Análisis A Priori es una evaluación matemática y universal sobre la lógica pura del algoritmo, mientras que la Prueba A Posteriori es una medición práctica y real pero limitada al entorno físico de la computadora donde se ejecuta el programa.

## 4. Sección Práctica 2: Iteración vs. Recursión
Un algoritmo recursivo debe llamarse a sí mismo resolviendo un problema menor e incluir siempre un "caso base" para evitar ciclos infinitos.

### [CÓDIGO 2.1] Sucesión de Fibonacci Recursiva
```python
# Fi = Fi-1 + Fi-2 para i >= 2; F0 = 0, F1 = 1
def fibonacci_recursivo(n):
    if n == 0: return 0
    elif n == 1: return 1
    else: return fibonacci_recursivo(n-1) + fibonacci_recursivo(n-2)

````python
def fibonacci(n):
    # Casos base: F(0) = 0, F(1) = 1
    if n <= 0:
        return 0
    elif n == 1:
        return 1

    # Caso recursivo: F(n) = F(n-1) + F(n-2)
    return fibonacci(n - 1) + fibonacci(n - 2)

# Ejemplo de uso
posicion = 7
print(f"El número Fibonacci en la posición {posicion} es: {fibonacci(posicion)}")
````


### [CÓDIGO 2.2] Factorial (Iterativo)
```python
# n! = 1 * 2 * 3 ... * n[cite: 2]
def factorial_iterativo(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

````python
def factorial(n):
    # El factorial no está definido para números negativos
    if n < 0:
        raise ValueError("El factorial no está definido para números negativos.")

    resultado = 1
    for i in range(2, n + 1):
        resultado *= i

    return resultado

# Ejemplo de uso
numero = 5
print(f"El factorial de {numero} es: {factorial(numero)}")
````

### Pregunta de Análisis 1 (Responde en una celda de texto):
Al analizar un algoritmo de ordenamiento sobre una lista, ¿qué escenarios representarían el "caso peor" y el "caso mejor" de los datos de entrada según la literatura revisada?

**RESPUESTA:**

Para la mayoría de los algoritmos de ordenamiento tradicionales (como inserción, burbuja o selección), los escenarios se definen de la siguiente manera:
Caso mejor: Ocurre cuando la lista de entrada ya está completamente ordenada en el sentido deseado. Bajo este escenario, algoritmos como el de inserción directa evitan realizar desplazamientos e intercambios, logrando un tiempo de ejecución lineal.
Caso peor: Ocurre cuando la lista de entrada está ordenada en sentido inverso (totalmente contraria al orden objetivo). Esto obliga al algoritmo a realizar el número máximo de comparaciones y movimientos de datos, resultando en una complejidad de tiempo cuadrática.
La excepción de Quicksort: En el algoritmo Quicksort, el peor y el mejor caso no dependen directamente de qué tan ordenada esté la lista inicial, sino del pivote que se elija para realizar las particiones. Si el pivote siempre se selecciona de manera desbalanceada (por ejemplo, eligiendo siempre el elemento mínimo o máximo), el rendimiento se degrada a cuadrático. Curiosamente, en implementaciones simples de Quicksort sin optimizaciones, una lista que ya está completamente ordenada puede llegar a representar su peor caso

````python
# Factorial Recursivo
def factorial_recursivo(n):
    if n < 0:
        raise ValueError("El factorial no está definido para números negativos.")
    if n <= 1:
        return 1
    return n * factorial_recursivo(n - 1)

# Fibonacci Iterativo
def fibonacci_iterativo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1

    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

# Ejemplos de uso
print(f"Factorial de 5: {factorial_recursivo(5)}")       # Salida: 120
print(f"Fibonacci(35): {fibonacci_iterativo(35)}")       # Salida: 9227465
````

````python
import time

# Definición de algoritmos
def fibonacci_recursivo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

def fibonacci_iterativo(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

n = 35

# Medición recursivo
inicio = time.time()
res_rec = fibonacci_recursivo(n)
fin = time.time()
tiempo_rec = fin - inicio

# Medición iterativo
inicio = time.time()
res_iter = fibonacci_iterativo(n)
fin = time.time()
tiempo_iter = fin - inicio

# Resultados
print(f"--- RESULTADOS PARA N = {n} ---")
print(f"Recursivo: Resultado = {res_rec} | Tiempo: {tiempo_rec:.6f} segundos")
print(f"Iterativo: Resultado = {res_iter} | Tiempo: {tiempo_iter:.6f} segundos")
````
